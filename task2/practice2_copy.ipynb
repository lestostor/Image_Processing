```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
from skimage.metrics import structural_similarity, mean_squared_error
```

```python
image = cv2.imread('sar_1.jpg')
image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
plt.imshow(image_gray, cmap="gray")
```

```python
# шум Гаусса
mean = 0
stddev = 100
noise_gauss = np.zeros(image_gray.shape, np.uint8)
cv2.randn(noise_gauss, mean, stddev)
image_noise_gauss = cv2.add(image_gray,noise_gauss)
plt.imshow(image_noise_gauss, cmap="gray")
```

```python
# постоянный шум
image = cv2.imread('sar_1.jpg', 0).astype(np.float32)
A = 50
noise_uniform = np.random.uniform(-A, A, image.shape).astype(np.float32)
image_noise_uniform = np.clip(image + noise_uniform, 0, 255)
image_noise_uniform = image_noise_uniform.astype(np.uint8)

plt.imshow(image_noise_uniform, cmap="gray")
```

```python
# медианный фильтр и шум гаусса
image_gauss_median_3 = cv2.medianBlur(image_noise_gauss, 3)
mse_gauss_median_3 = mean_squared_error(image_gray, image_gauss_median_3)
(ssim_gauss_median_3, diff) = structural_similarity(image_gray, image_gauss_median_3, full=True)

plt.imshow(image_gauss_median_3, cmap="gray")
print("MSE: ", mse_gauss_median_3)
print("SSIM: ", ssim_gauss_median_3)
```
