```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
from skimage.metrics import structural_similarity, mean_squared_error
```

```python
# №1
image = cv2.imread('sar_1_gray.jpg')
plt.imshow(image)
```

```python
# №2
histSize = 256
histRange = (0, 256)
accumulate = False

b_hist = cv2.calcHist([image], [0], None, [histSize], histRange, accumulate=accumulate)
plt.plot(b_hist)
```

```python
# №3
def gamma_correct(img, gamma_pow) :
    img = img / 255.0
    result = np.power(img, gamma_pow)
    return (result * 255).astype(np.uint8)

gamma1 = gamma_correct(image, 0.5)
gamma2 = gamma_correct(image, 2.0)
```

```python
# №4
# сравнение со светлой версией
(ssim, diff) = structural_similarity(image, gamma1, full=True, win_size = 3)
mse = mean_squared_error(image, gamma1)
diff = (diff * 255).astype("uint8")
print("SSIM: {}".format(ssim))
print("MSE: ", mse)
plt.imshow(diff)
```

```python
# сравнение с тёмной версией
(ssim, diff) = structural_similarity(image, gamma2, full=True, win_size = 3)
mse = mean_squared_error(image, gamma2)
diff = (diff * 255).astype("uint8")
print("SSIM: {}".format(ssim))
print("MSE: ", mse)
plt.imshow(diff)
```

```python
# №5
img_gray = cv2.imread('sar_1_gray.jpg', 0)
eq_gray = cv2.equalizeHist(img_gray)

E_t = img_gray.mean()
D_t = img_gray.std()
E_s = eq_gray.mean()
D_s = eq_gray.std()

image_corrected = (E_s + (img_gray - E_t) * (D_s / D_t)).clip(0, 255)
plt.imshow(image_corrected, cmap="gray")
```

```python
types = [cv2.THRESH_BINARY, cv2.THRESH_BINARY_INV, cv2.THRESH_TRUNC, cv2.THRESH_TOZERO, cv2.THRESH_TOZERO_INV]
values = [50, 100, 150, 200]

for type in types:
    for value in values:
        _,thresh = cv2.threshold(img_gray,value,255,type)
        plt.imshow(thresh, cmap='gray')
        plt.show()
```
